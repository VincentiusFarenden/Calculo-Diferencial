### Función ReLu

Sea una función ReLu definida como:
$$f(x)=max(0,x)$$

O expresada por tramos:

$$ReLU(x)=\cases{0 \hspace{1cm}\text{ si } x<0\\x\hspace{1cm} \text{ si } x\geq 0}$$

Puede declararse de forma simbólica como:

```python
x = sp.Symbol('x')                                          #Declara un símbolo (x)

f_relu_sym = sp.Lambda(x, sp.Max(0,x))                      #Se define la función

f_relu = sp.lambdify(x, f_relu_sym, 'numpy')                #Convierte una función simbólica (SymPy) en una numérica (NumPy)
```

## **Composición de funciones, límites y continuidad**


### Composición de funciones

```python
x = sp.Symbol('x')                              #Declara un símbolo (x)

f = sp.Lambda(x, EXPRESIÓN_F)                   #Define una función simbólica
g = sp.Lambda(x, EXPRESIÓN_G)                   #Define una función simbólica

fog = sp.Lambda(x, f(g(x)))                     #f(g(x))
gof = sp.Lambda(x, g(f(x)))                     #g(f(x))

fog_simplificada = sp.simplify(fog(x))          #Simplifica la expresión
gof_simplificada = sp.simplify(gof(x))          #Simplifica la expresión

print(f'f(g(x)) = {fog_simplificada}')
print(f'g(f(x)) = {gof_simplificada}')
```
