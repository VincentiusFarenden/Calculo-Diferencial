### Problema 3 💻/🐍

Durante los 15 primeros días de una investigación, la población de una colonia de bacterias está dada por:

$$P(t) = \frac{5.000}{1 + 20e^{-0,6t}}$$

a) Grafique la función y describa su comportamiento en términos de crecimiento y decrecimiento.


b) Calcule y grafique $P'(t)$. ¿Es coherente con lo visto en el gráfico de $P(t)$?

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

t=sp.Symbol('t')
P=sp.Lambda(t, 5000/ (1+20*sp.exp(-0.6*t)))
p_num=sp.lambdify(P.variables,P.expr,'numpy')
t_num=np.linspace(0,15,100)

fig,ax=plt.subplots(figsize=(6,4))
ax.plot(t_num,p_num(t_num))
ax.set_title('Funcion P(t)')
ax.set_xlabel('Poblacion de bacterias')
ax.set_ylabel('Tiempo en dias')
plt.show()
```

c) Interprete el significado de $P'(t)$ en el contexto biológico.

La derivada muestra que inicia hasta el dia 3 aproximadamente, la población de bacterias tiene un crecimiento (velocidad) más rápido, luego la velocidad baja bruscamente, y se comienza a estabilizar a partir del día 10.

### Problema 4 💻/🐍

La posición (en metros) de una partícula que se mueve en línea recta, durante los 20 primeros segundos, está dada por:

$$s(t) = t^3 e^{-0,3t} + 2t$$

donde $t$ corresponde al tiempo medido en segundos.

a) Grafique en el dominio contextualizado las funciones $s(t)$, $s'(t)$ y $s''(t)$.

```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

t=sp.Symbol('t')
S=sp.Lambda(t, t**3*sp.exp(-0.3*t)+2*t)

s_num=sp.lambdify(t,S(t),'numpy')
t_num=np.linspace(0,20,100)

#Creación del gráfico y eje
fig,ax=plt.subplots(figsize=(6,4))
ax.plot(t_num,s_num(t_num))

#Títulos del gráfico
ax.set_title('Derivada P(t)')
ax.set_ylabel('Población de bacterias/día')
ax.set_xlabel('TIempo en días')

plt.grid()
plt.show()

display(S)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

t=sp.Symbol('t')
S=sp.Lambda(t, t**3*sp.exp(-0.3*t)+2*t)
ds=sp.diff(S(t),t)

ds_num=sp.lambdify(t,ds,'numpy') #función derivada sympy->numpy

t_num=np.linspace(0,20,100)

#Creación del gráfico y eje
fig,ax=plt.subplots(figsize=(6,4))
ax.plot(t_num,ds_num(t_num))

#Títulos del gráfico
ax.set_title('Gráfico de la derivada s(t)')
ax.set_ylabel('Posición/tiempo (metros/segundo)')
ax.set_xlabel('TIempo en segundos')
ax.set_xlim(0,20)
ax.set_xticks(np.arange(0,21,1))

plt.grid()
plt.show()

display(S)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

t=sp.Symbol('t')
S=sp.Lambda(t,t**3*sp.exp(-0.3*t)+2*t)
ds=sp.diff(S(t),t)

dds=sp.diff(ds,t)
display(dds)

dds_num=sp.lambdify(t,dds,'numpy')

t_num=np.linspace(0,20,100)

#Creación del grafico y ejes
fig,ax=plt.subplots(figsize=(6,4))
ax.plot(t_num,dds_num(t_num))

#Titulos del gráfico
ax.set_title('Gráfico de la segunda derivada s(t)')
ax.set_ylabel('Posición/tiempo')
```
