#Definición de variable

```python
import sympy as sp #Importación de librería necesaria
x=sp.symbols('x')
```

#Polinomios

```python
x,n= sp.symbols('x n')
f=x**(n-1)-x**n+x**(n+2)
display(f)
```

#Exponencial

```python
h=sp.exp(x) #Llamo a la función exponencial
display(h)
```

#Función exponencial con base "a" (cualquier número)

```python
a=sp.symbols('a')
g=a**x
display(g)
```

#Función logaritmo natural

```python
j=sp.log('x') #Llamo a la función logaritmo natural
display(j)
```

#Base diferente

```python
k=sp.log(x,10) #Dentro del argumento va la "x" y la base "número"
display(k)
```

#Evaluar funciones para un valor de "x"

```python
import sympy as sp
x=sp.symbols('x')
m=sp.Lambda(x,x**2)
display(m)
display(m(3))
```

#Resolver ecuaciones

```python
import sympy as sp
x=sp.symbols('x')
f=x**2-5*x+6

display(f)
```

```python
ecuacion=sp.Eq(f,0) #Se define la ecuación para igualarla a 0
display(ecuacion)
```

```python
solucion=sp.solve(ecuacion) #Resuelve la ecuación y encuentra las soluciones
display(solucion)
```
