### Problema 12 📝



Dadas las siguientes funciones $f(x)$ y $g(x)$, determine $h(x)=(f\circ g)(x)$ y $h'(x)$

a) $$\begin{aligned}
f(x)&=e^x\\
g(x)&=4x\\
\end{aligned}$$

<details>
<summary><b>&#9654; Ver solución</b></summary>
<br>

$f(x)=e^x$, $g(x)=4x$

$h(x)=(f\circ g)(x)=f(g(x))=e^{4x}$

Regla de la cadena (exterior $e^{\square}$, interior $4x$, derivada interna $4$):

$$h'(x) = 4e^{4x}$$

</details>

b) $$\begin{aligned}
f(x)&=log(x)\\
g(x)&=2x+5
\end{aligned}$$

<details>
<summary><b>&#9654; Ver solución</b></summary>
<br>

$f(x)=\log(x)$, $g(x)=2x+5$

$h(x)=(f\circ g)(x)=f(g(x))=\log(2x+5)$

Regla de la cadena con $\dfrac{d}{dx}\log(u)=\dfrac{u'}{u\ln 10}$, $u=2x+5$, $u'=2$:

$$h'(x) = \dfrac{2}{(2x+5)\ln 10}$$

</details>
