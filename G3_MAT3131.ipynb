### Problema 15 💻/🐍

$$p(t)=0,005e^{0,75t}+12$$

```python
import numpy as np
import matplotlib.pyplot as plt

def p(t):
    return 0.005*np.exp(0.75*t)+12

t=np.linspace(0,12,100)

fig,ax=plt.subplots(figsize=(6,4))

ax.plot(t,p(t))

ax.set_title('Función de población respecto al tiempo')
ax.set_xlabel('Tiempo en años')
ax.set_ylabel('Población en miles de habitantes')

plt.show()
```
